```python
# Python Library Management System - Completed

# Install required libraries (already available in Colab)
!pip install requests -q

import pickle
import sqlite3
import requests

# 1. Data Types and Variables
book_title = "Python Programming"
book_quantity = 5
book_price = 29.99
is_available = True

print(f"Book: {book_title}, Quantity: {book_quantity}, Price: ${book_price}, Available: {is_available}")
```

```python
# 2. Lists
books = ["Python Basics", "Advanced Python", "Data Science with Python"]

# TODO: Add two more books to the list
books.append("Machine Learning with Python")          # Add at the end
books.insert(1, "Python for Data Analysis")           # Insert at specific position (index 1)

print("Updated book list:", books)
```

```python
# 3. Loops and range()
print("\nAvailable books with indices:")
# TODO: Use a for loop and range() to print books with their indices
for i in range(len(books)):
    print(f"{i + 1}. {books[i]}")
```

```python
# 4. User Input
print("\n--- Adding a new book via user input ---")
new_book = input("Enter a new book title: ").strip()
if new_book:  # Avoid adding empty strings
    books.append(new_book)
else:
    print("No book title entered. Skipping addition.")

print("Updated book list after user input:", books)
```

```python
# 5. Functions and Conditional Statements
def check_availability(book_name):
    """
    Check if a book is available in the library list.
    Handles case-insensitive comparison and strips whitespace.
    """
    if not book_name or not isinstance(book_name, str):
        return False

    book_name_clean = book_name.strip().lower()

    # Check if book exists in the list (case-insensitive)
    for book in books:
        if book.strip().lower() == book_name_clean:
            return True
    return False

# Test the function
print(f"\nIs 'Python Basics' available? {check_availability('Python Basics')}")
print(f"Is 'python basics' available? {check_availability('python basics')}")  # Case insensitive test
print(f"Is 'Nonexistent Book' available? {check_availability('Nonexistent Book')}")
```

```python
# 6. File I/O

def save_books_to_file(filename):
    """Save the current books list to a text file, one book per line."""
    try:
        with open(filename, 'w', encoding='utf-8') as f:
            for book in books:
                f.write(f"{book}\n")
        print(f"Books successfully saved to {filename}")
    except IOError as e:
        print(f"Error saving to file: {e}")

def read_books_from_file(filename):
    """
    Read books from a text file and return them as a list.
    Handles file not found and other I/O errors.
    """
    try:
        with open(filename, 'r', encoding='utf-8') as f:
            loaded_books = [line.strip() for line in f.readlines() if line.strip()]
        print(f"Successfully loaded {len(loaded_books)} books from {filename}")
        return loaded_books
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return []
    except IOError as e:
        print(f"Error reading file: {e}")
        return []

# Test file I/O
save_books_to_file("books.txt")
print("\nContents of books.txt:")
!cat books.txt

# Read back to verify
loaded_from_file = read_books_from_file("books.txt")
print("Books loaded from file:", loaded_from_file)
```

```python
# 7. Pickle for serialization

def save_books_pickle(filename):
    """Save books list to a binary pickle file."""
    try:
        with open(filename, 'wb') as f:
            pickle.dump(books, f)
        print(f"Books successfully saved to pickle file: {filename}")
    except Exception as e:
        print(f"Error saving pickle file: {e}")

def load_books_pickle(filename):
    """
    Load books from a pickle file.
    Returns empty list if file doesn't exist or loading fails.
    """
    try:
        with open(filename, 'rb') as f:
            loaded_books = pickle.load(f)
        print(f"Successfully loaded {len(loaded_books)} books from pickle file.")
        return loaded_books
    except FileNotFoundError:
        print(f"Error: Pickle file '{filename}' not found.")
        return []
    except (pickle.PickleError, EOFError) as e:
        print(f"Error loading pickle file (corrupted?): {e}")
        return []
    except Exception as e:
        print(f"Unexpected error loading pickle: {e}")
        return []

# Test pickle functions
save_books_pickle("books.pkl")
loaded_from_pickle = load_books_pickle("books.pkl")
print("Books loaded from pickle:", loaded_from_pickle)
```

```python
# 8. SQLite Database Interaction

def create_books_table():
    """Create the books table if it doesn't exist."""
    try:
        conn = sqlite3.connect('library.db')
        cursor = conn.cursor()
        cursor.execute('''CREATE TABLE IF NOT EXISTS books
                          (id INTEGER PRIMARY KEY,
                           title TEXT NOT NULL,
                           quantity INTEGER DEFAULT 0)''')
        conn.commit()
        conn.close()
        print("Books table created/verified successfully.")
    except sqlite3.Error as e:
        print(f"SQLite error creating table: {e}")

def add_book_to_db(title, quantity):
    """
    Add a book to the SQLite database.
    Validates input and handles database errors.
    """
    if not title or not isinstance(title, str):
        print("Error: Book title must be a non-empty string.")
        return
    if not isinstance(quantity, int) or quantity < 0:
        print("Error: Quantity must be a non-negative integer.")
        return

    try:
        conn = sqlite3.connect('library.db')
        cursor = conn.cursor()

        cursor.execute('''INSERT INTO books (title, quantity)
                          VALUES (?, ?)''', (title.strip(), quantity))

        conn.commit()
        print(f"Book '{title}' added to database with quantity {quantity}.")
    except sqlite3.Error as e:
        print(f"Database error adding book: {e}")
    finally:
        conn.close()

# Test database functions
create_books_table()
add_book_to_db("Python for Beginners", 10)
add_book_to_db("Clean Code", 5)  # Adding another book for demonstration

# Optional: View all books in DB
def show_all_books_db():
    try:
        conn = sqlite3.connect('library.db')
        cursor = conn.cursor()
        cursor.execute("SELECT * FROM books")
        rows = cursor.fetchall()
        print("\nBooks in SQLite Database:")
        for row in rows:
            print(f"ID: {row[0]}, Title: {row[1]}, Quantity: {row[2]}")
        conn.close()
    except sqlite3.Error as e:
        print(f"Error reading database: {e}")

show_all_books_db()
```

```python
# 9. Web API Interaction (GitHub API)
def get_python_repos():
    """Fetch top 5 Python repositories from GitHub by stars."""
    url = "https://api.github.com/search/repositories"
    params = {"q": "language:python", "sort": "stars", "order": "desc", "per_page": 5}

    try:
        response = requests.get(url, params=params, timeout=10)
        response.raise_for_status()  # Raise error for bad status codes
        data = response.json()

        if 'items' not in data:
            print("Unexpected API response format.")
            return []

        repo_names = [repo['full_name'] for repo in data['items']]
        return repo_names
    except requests.exceptions.Timeout:
        print("Error: Request timed out.")
        return []
    except requests.exceptions.RequestException as e:
        print(f"Error fetching GitHub repositories: {e}")
        return []
    except Exception as e:
        print(f"Unexpected error: {e}")
        return []

# Test API function
print("\nTop 5 Python repositories on GitHub:")
top_repos = get_python_repos()
for i, repo in enumerate(top_repos, 1):
    print(f"{i}. {repo}")

print("\n✅ Library Management System operations completed successfully!")
```
